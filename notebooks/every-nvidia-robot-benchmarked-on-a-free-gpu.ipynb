## The question

NVIDIA's Isaac asset library ships robots from Boston Dynamics, Unitree,
ANYbotics, Kuka, Fanuc, Shadow Robot, NASA and 43 other vendors. There is no
published answer to two obvious questions:

1. **Which of them actually load and simulate?**
2. **How fast, on hardware you can get for nothing?**

This notebook answers both. Every number was measured on Kaggle's free Tesla
T4 - physics only, since the RTX renderer does not initialise on that tier.

**Result: 49 of 50 vendors work.** A free GPU runs almost the entire library
faster than real time.

````python
import matplotlib.pyplot as plt

dof = np.array([r["num_dof"] for r in ok])
rt  = np.array([r["realtime_factor"] for r in ok])
names = [r["vendor"] for r in ok]

fig, ax = plt.subplots(1, 2, figsize=(14, 5))

ax[0].scatter(dof, rt, s=52, color="#2a9d8f", edgecolor="#1d5c56", zorder=3)
ax[0].axhline(1.0, color="#c1443b", ls="--", lw=1.5, label="real time")
for n, x, y in zip(names, dof, rt):
    if x >= 22 or y < 1.4:
        ax[0].annotate(n, (x, y), textcoords="offset points", xytext=(6, 3),
                       fontsize=8, color="#31414f")
c = np.polyfit(np.log(dof), np.log(rt), 1)
xs = np.linspace(dof.min(), dof.max(), 50)
r_corr = np.corrcoef(np.log(dof), np.log(rt))[0, 1]
ax[0].plot(xs, np.exp(np.polyval(c, np.log(xs))), ":", color="#8d99ae", lw=1.8,
           label=f"trend (r={r_corr:.2f})")
ax[0].set_yscale("log")
ax[0].set_xlabel("degrees of freedom"); ax[0].set_ylabel("realtime factor")
ax[0].set_title("Simulation speed vs robot complexity")
ax[0].grid(alpha=.3, which="both"); ax[0].legend(loc="lower left", fontsize=9)

order = np.argsort(-dof)[:20]
ax[1].barh([names[i] for i in order][::-1], [dof[i] for i in order][::-1],
           color="#264653")
ax[1].set_xlabel("degrees of freedom")
ax[1].set_title("20 most complex robots in the library")
ax[1].grid(alpha=.3, axis="x")
plt.tight_layout(); plt.show()

print(f"log-log correlation, DOF vs speed: {r_corr:.3f}")
print(f"below real time: {[n for n,y in zip(names,rt) if y < 1.0]}")
````

Complexity predicts cost, as it must: **r = -0.82** on log-log. Only the two
77-DOF humanoids (Galbot, SanctuaryAI) drop below real time. A 12-DOF quadruped
like Spot or ANYmal runs 8-9x faster than real time on a **free** GPU, which is
comfortably enough for RL data collection.

That correlation is also the sanity check. An earlier version of this benchmark
produced a table where a 9-DOF arm appeared 5.8x *slower* than a 33-DOF
humanoid - see the last section for why, because the reason is the most useful
thing here.

## Trap 1 — Isaac Sim 6.0 moved every robot

The flat layout is gone. Robots are grouped by **manufacturer** now:

```
/Isaac/Robots/Franka/franka.usd                        404
/Isaac/Robots/FrankaRobotics/FrankaPanda/franka.usd     OK
```

Every tutorial written before 6.0 uses the old path. The error is a plain
`FileNotFoundError`, which reads like a broken install rather than a moved file.

## Trap 2 — not every robot is a `.usd`

USD has several serialisations. Some vendors ship **`.usda`** (the ASCII form)
as their canonical asset:

```
Mecademic/meca500/meca500.usda
InspireRobots/Inspire_Hand_RH56DFX_right/inspire_hand.usda
```

A `glob("*.usd")` misses these entirely. When I filtered that way, two working
robots were recorded as broken assets and my catalogue undercounted the library
by **134 files**. Count `.usd`, `.usda`, `.usdc` and `.usdz`.

## Trap 3 — `World.clear_instance()` does not free the robot

This is the one that produced fake numbers.

Loading robots back-to-back in one process leaves every previous robot resident
on the stage. Throughput then measures **load order**, not the robot:

```
robot  1  1X Neo, 33 DOF    107.3 steps/s
robot  8  Booster,  12 DOF    29.8 steps/s
robot 16  Franka,    9 DOF    18.5 steps/s     <- simplest robot, slowest result
          ...SIGSEGV at robot 17
```

Every number was plausible. The table looked fine. It was meaningless, and the
giveaway was that DOF and speed had the wrong relationship - the same
correlation plotted above.

**The fix:** a fresh stage per robot (`omni.usd.get_context().new_stage()`) and
a fresh process per chunk of 8. Chunked rather than one-process-per-robot
because Kit startup costs about 25 s; 49 launches would spend 20 minutes just
booting.

Each chunk writes results to disk before exiting, so a crash costs 8 robots
rather than the whole run.

## The one real failure

**NTNU ARL-Robot-1** raises `ValueError: zero-size array to reduction operation`
on `art.num_dof`. Its asset has no articulation joints at all - a property of
the asset, not a loader problem.

Everything else in the library works. If a robot fails for you, check the three
traps above before concluding the asset is broken; all four of my initial
failures turned out to be my own filtering.

## Using this

The linked dataset has the full table plus `robot_catalogue.json` - all
**2,034 robot assets across 50 vendors**, so you can pick by vendor without
crawling S3 yourself.

Practical notes:

- **Pick by DOF for your compute budget.** Under ~25 DOF you get 2-9x real
  time on a free T4. Above 50 DOF you approach or drop below it.
- **Prefer the file whose name matches its folder** (`anymal_b/anymal_b.usd`).
  That is NVIDIA's convention for the canonical asset; the `payloads/` and
  `HighResProps/` directories hold meshes and props, not robots.
- **Rendering is unavailable on Kaggle's free tier**, so this is physics only.
  Everything here applies to state-based RL, control, and dynamics work.

## Reproducing this

Every notebook in this series runs on free infrastructure. Nothing here needs
a paid GPU.

| Method | Free allowance | Best for |
|---|---|---|
| Kaggle | 30 GPU hr/week, 2x T4 | Running this notebook as-is |
| Lightning AI | 80 GPU hr/month, persistent disk | Heavy Isaac Sim generation |
| Google Colab | best-effort T4 | Quick smoke tests |
| NVIDIA DLI | free hosted labs | Learning the Isaac Sim GUI |

**The one gotcha worth remembering:** Isaac Sim needs **RT cores**. A T4, L4,
L40S or any RTX card is fine. An **A100 or H100 is not** — those have no RT
cores, so the RTX renderer is unsupported or unusably slow. It is the most
counterintuitive constraint in cloud robotics simulation, and it bites people
who assume the more expensive GPU must be the better one.
